## Data Profiling Findings

After profiling the `order_items` dataset, the following findings and transformation steps were identified.

## 1. Column Names

The current column names are not fully consistent with the naming convention used in the project.

**Action:**

* Rename columns to follow a consistent `snake_case` naming convention.
* Rename `id` to `product_id`.
* Rename `order_userId` to `user_id`.
* Rename `discountPercentage` to `discount_percentage`.
* Rename `discountedTotal` to `discounted_total`.

## 2. Unnecessary Columns

The `thumbnail` column contains product image URLs and is not required for the current analytical use case.

**Action:**

* Remove the `thumbnail` column.

## 3. Data Types

The `order_id` and `user_id` columns are currently stored as `object`.

**Action:**

* Convert `order_id` and `user_id` to integer data types.

## 4. Missing Values

No missing values were found in the current API response.

**Action:**

* No missing-value handling is required for the current dataset.
* Missing-value handling can be added to the transformation logic to handle future API responses containing missing data.

## 5. Duplicate Records

Duplicate order-item records were identified during profiling and should be reviewed during the cleaning stage.

**Action:**

* Review duplicate records based on the appropriate order-item key before removing them.

## 6. Value Validation

The numerical columns were reviewed using descriptive statistics and value ranges.

* `quantity` contains positive values.
* `price` contains positive values.
* `discount_percentage` values are within the expected range.
* `total` and `discounted_total` contain positive values.

High-value records were observed, but they are consistent with the corresponding `price` and `quantity` values.

**Action:**

* Keep the high-value records.
* No outlier removal will be applied based on the current profiling results.

## 7. Transformation Summary

The following transformations will be applied to the dataset:

1. Rename columns to follow a consistent naming convention.
2. Remove the `thumbnail` column.
3. Convert `order_id` and `user_id` to integer data types.
4. Review and handle duplicate order-item records.
5. Validate numerical values and business rules.
6. Keep high-value records as they are not considered invalid based on the current profiling.
